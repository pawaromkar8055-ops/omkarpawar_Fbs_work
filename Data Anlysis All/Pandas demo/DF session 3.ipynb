### Change values There are two methods 
## 1 = loc methods, use for location , ID, SALARY , NAME use loc ,by using this menthods we change values in orignal also
## 2 = iloc methods, we use changing by index ,by using this menthods we change values in orignal also

```python
data.loc[data['SALARY'] < 35000, 'SALARY'] = 35000
data
```

```python
data.loc[[0,2], ['NAME', 'SALARY']]
```

```python
data.loc[[0,2], 'SALARY']
```
