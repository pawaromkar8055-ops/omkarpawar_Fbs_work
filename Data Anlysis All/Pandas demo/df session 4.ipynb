```python
emp = pd.DataFrame({'ID':[101,102,103,104],
                    'NAME':['Rohit','Virat','Kishor','Avi'],
                    'AGE':[50,40,60,70],
                   'DID':[1, 3, 1, 5]})
emp
```

```python
dept = pd.DataFrame({'DID': [1,2,3,4], 
                    'DNAME':['IT', 'Admin', 'HR', 'Sales']})
dept
```

## What merge 
## Joinng two differnt data frames for our useful information , into that one similar colume avalabe in that dataset, how is our parameter name 
1st method = inner methods = return match record form the both dataframe.....
2nd method = left methods = return match record from both dataframe  and unmatch record from left dataframe.....
3rd method = right methods = return match record from both dataframe  and unmatch record from right dataframe......
4th method = outer methods = return all match record both dataframe and unmatch record from left dataframe , unmatch record from right dataframe

```python
pd.merge(emp, dept)
```

```python
dept2 = pd.DataFrame({'ID': [1,2,3,4], 
                    'DNAME':['IT', 'Admin', 'HR', 'Sales']})
dept2
```

If common caloume name from both dataframe  is not same use parameter lift_on and right_on when use colume name is not same
bydefult inner

```python
pd.merge(emp,dept2,left_on='DID',right_on='ID')
```

```python
pd.merge(emp,dept, how='left' )
```

```python
pd.merge(emp,dept, how='right' )
```

```python
pd.merge(emp,dept, how='outer' )
```
